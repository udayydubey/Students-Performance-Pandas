### Maximum weekly self-study hours

```python
hours = df['weekly_self_study_hours'].max()
df[df['weekly_self_study_hours'] == hours]
```

### Highest attendance and class participation

```python
attendance = df['attendance_percentage'].max()
df[df['attendance_percentage'] == attendance]

participation = df['class_participation'].max()
df[df['class_participation'] == participation]
```

### Score classification

```python
def level(x):
    if x >= 90:
        return 'High'
    elif x >= 70:
        return 'Medium'
    else:
        return 'Low'

df['ScoringLevel'] = df['total_score'].apply(level)
```

```python
max_score = df['total_score'].max()
df[df['total_score'] == max_score]

len(df[df['total_score'] == 100])
```

### Derived metrics

```python
df['efficiency'] = df['total_score'] / df['weekly_self_study_hours']
```

```python
def category(x):
    if x < 10:
        return 'Light'
    elif x <= 20:
        return 'Moderate'
    else:
        return 'Heavy'

df['Category'] = df['weekly_self_study_hours'].apply(category)
```
